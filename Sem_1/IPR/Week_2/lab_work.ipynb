Exercise 2

```python
import torch
```

```python
x = torch.tensor([-10.0, 10.0, 8.0])
x.requires_grad_(True)
print(f"X values: {x}")
print()
```

```python
def sigmoid(x):
    return 1 / (1+ torch.exp(-x))

print(f"Sigmoid of X: {sigmoid(x)}")
print()

y_sigmoid = sigmoid(x)

y_sigmoid.sum().backward()

print(f"Sigmoid Derivative of X: {x.grad}")
print()
```

```python
print("Sigmoid Function Values:")
for i, x_val in enumerate(x):
    print(f"sigmoid({x_val.item()}) = {y_sigmoid[i].item()}")
print()

print("Derivative of Sigmoid Function Values:")
for i, x_val in enumerate(x):
    print(f"sigmoid'({x_val.item()}) = {x.grad[i].item()}")
```

```python
import torch
```

```python
A = torch.tensor([[1,2,-3], [4,5,10]])
print("Matrix A:")
print(A)
print()

B = torch.tensor([[10,11,-2], [13, -3, 8]])
print("Matrix B:")
print(B)
print()
```

```python
product = A*B
print("Element-wise Product of A and B:")
print(product)
print()
```

```python
a = 10 
b_transpose = B.T
matrix_addition = a + b_transpose
print("Matrix Addition (A + B^T):")
print(matrix_addition)
```

```python
print("Matrix A Transpose:")
print(A.T)
print()

print("Matrix B Transpose:")
print(B.T)
print()
```

```python
A = torch.tensor([[1,2,3], [4,5,6]], dtype=torch.float32)
print(A)
print()

B = torch.tensor([[10,11], [-2,0], [1, -3]], dtype=torch.float32)
print(B)
print()
```

```python
product = torch.matmul(A, B)
print("Matrix Product of A and B:")
print(product)
```
